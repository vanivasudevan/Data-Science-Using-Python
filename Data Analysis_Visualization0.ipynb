```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
# Step 1: Data Cleaning and Preprocessing
# Load sample dataset
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', np.nan],
    'Age': [25, 30, np.nan, 35, 40, 29, np.nan, 50],
    'Salary': [50000, 60000, 70000, np.nan, 90000, 62000, 73000, 80000],
    'City': ['New York', 'los Angeles', 'New York', 'Chicago', np.nan, 'chicago', 'Los Angeles', 'New York']
})
df
```

```python
# Handling missing values
df['Age'].fillna(df['Age'].median(), inplace=True)  # Filling missing Age with median
df['Salary'].fillna(df['Salary'].mean(), inplace=True)  # Filling missing Salary with mean
df.dropna(subset=['Name', 'City'], inplace=True)  # Dropping rows with missing Name or City

# Ensuring data consistency
df['City'] = df['City'].str.title() # Standardizing city names
print("Cleaned Data:")
print(df)
```

```python
# Step 2: Statistical Analysis
print("\nDescriptive Statistics:")
print(df.describe(include='all'))
```

```python
# Step 3: Exploratory Data Analysis (EDA)
plt.figure(figsize=(12, 6))
sns.histplot(df['Age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(8, 6))
numeric_corr = df.corr(numeric_only=True)
sns.heatmap(numeric_corr, annot=True, linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Boxplot for Salary distribution
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Salary'])
plt.title("Salary Distribution")
plt.show()
```
